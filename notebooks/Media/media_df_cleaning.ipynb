```python
monthly_sentiment = (
    Sentiment.groupby('year_month')
    .agg(
        mean_title_positive=('title_positive','mean'),
        mean_title_neutral=('title_neutral','mean'),
        mean_title_negative=('title_negative','mean'),
        mean_title_sentiment=('title_sentiment', 'mean'),
        mean_body_positive=('body_positive','mean'),
        mean_body_neutral=('body_neutral','mean'),
        mean_body_negative=('body_negative','mean'),
        mean_title_risk=('title_risk', 'mean'),
        mean_body_sentiment=('body_sentiment', 'mean'),
        mean_body_risk=('body_risk', 'mean'),
        article_count=('year_month', 'size')
    )
    .reset_index()
)
```

```python
MonthlyMedia = monthly_sentiment.sort_values('year_month').copy()

body_cols = [
    'mean_title_positive', 'mean_title_neutral',
       'mean_title_negative', 'mean_title_sentiment', 'mean_body_positive',
       'mean_body_neutral', 'mean_body_negative', 'mean_title_risk',
       'mean_body_sentiment', 'mean_body_risk', 'article_count'
]

for col in body_cols:
    past_mean = MonthlyMedia[col].shift(1).expanding(min_periods=1).mean()
    MonthlyMedia[col] = MonthlyMedia[col].fillna(past_mean)
```
